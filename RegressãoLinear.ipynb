```python
from sklearn.linear_model import LinearRegression

#atributos (X)
horas_estudo = [[1],[2],[3],[4],[5]]   #tabela

#target (Y)
pontuacao_teste = [50,55,65,70,80]

model = LinearRegression()
model.fit(horas_estudo, pontuacao_teste)


predcao = model.predict([[7.8]])

print(predcao)
```
