Atividade regressão linear bicicletas

```python
from sklearn.linear_model import LinearRegression

duracoes = [[10],[15],[20],[25],[30],[35],[40],[50],[55],[60]]

precos = [10.00,13.00,14.00,18.00,19.00,23.00,24.00,30.00,32.00,34.00]

model = LinearRegression()
model.fit(duracoes, precos)


previsao_45 = model.predict([[45]])
previsao_20 = model.predict([[20]])
previsao_60 = model.predict([[60]])


print(f"Preço estimado para 45 minutos: R$ {previsao_45[0]:.2f}")

print()

print(f"Previsão para 20 minutos: R$ {previsao_20[0]:.2f}")
print(f"Preço histórico para 20 minutos: R$ {14.00:.2f}")

print()

print(f"Previsão para 60 minutos: R$ {previsao_60[0]:.2f}")
print(f"Preço histórico para 60 minutos: R$ {34.00:.2f}")
```

No caso de 20 minutos:

Preço registrado: R$ 14,00
Previsão: R$ 14,65


No caso de 60 minutos:

Preço registrado: R$ 34,00
Previsão: R$ 34,56

### 2. Qual preço o modelo estimou para 45 minutos?


RESPOSTA: 
O modelo estimou R$ 26,56 para um aluguel de 45 minutos

RESPOSTA:

o modelo estimou R$ 14,65 e o registro histórico é de R$ 14,00

Para 60 minutos
estimou R$ 34,56 e o registro histórico é de R$ 34,00

### 4. Por que o preço previsto deve ser apresentado como uma estimativa? 

RESPOSTA: Porque é uma estimativa baseada em dados históricos e o preço real pode variar devido a descontos e outros fatores.


### 5. Você teria a mesma confiança em uma previsão para 300 minutos? Justifique considerando os dados disponíveis.

RESPOSTA: Não. A previsão para 300 minutos seria menos confiável, pois está muito além dos dados disponíveis, que vão até 60 minutos.

### Parecer — Pedal Livre

Com base nos registros históricos de aluguel, foi criado um modelo de regressão linear que relaciona a duração do aluguel ao preço pago.

Para um aluguel de 45 minutos, o modelo estima um preço de aproximadamente R$ 26,56.

As previsões para 20 e 60 minutos ficaram próximas dos preços registrados, mostrando que o modelo conseguiu representar bem a tendência geral dos dados. Entretanto, o valor deve ser considerado uma estimativa, pois os preços históricos apresentam pequenas variações.

A previsão é mais confiável para durações próximas das observadas nos registros, que variam de 10 a 60 minutos. Para períodos muito maiores, como 300 minutos, seria necessário ter mais dados históricos antes de confiar no resultado.

Conclusão: para um cliente que pretende alugar a bicicleta por 45 minutos, recomendamos considerar R$ 26,56 como preço estimado, sabendo que o valor real pode apresentar alguma diferença.
